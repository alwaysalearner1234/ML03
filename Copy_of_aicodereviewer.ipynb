```python
from transformers import pipeline

reviewer = pipeline(
    "text-generation",
    model="microsoft/phi-2"
)
```

```python
def review_code(code):

    prompt = f"""
Review the following code.

Give:
1. Bugs
2. Time Complexity
3. Improvements
4. Better Version

Code:
{code}
"""

    result = reviewer(
        prompt,
        max_new_tokens=200
    )

    return result[0]["generated_text"]
```

```python
import gradio as gr
```

```python
demo = gr.Interface(
    fn=review_code,
    inputs=gr.Textbox(
        lines=20,
        label="Paste Code"
    ),
    outputs="text",
    title="AI Code Reviewer"
)

demo.launch()
```
