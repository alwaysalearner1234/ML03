```python
# divide the data into input(x)features and output(y) labels/target

# input is always 2 dimensional; and output is always 1 dimensional

# in the column slicing part, if there is ':',then the array is 2 dimensional

#syntax :  var.iloc[row slicing,column slicing].values

x=df.iloc[0:6,0:1].values
x
```

```python
x.ndim
```

```python
#in the column slicing part, if there is no colon , the arrAY IS ONE DIMENSIONAL

y=df.iloc[0:6,1].values
y
```

```python
#6 import a classifier, regressor, or a classifier

from sklearn.linear_model import LinearRegression
model=LinearRegression()
```

```python
#7 fit the model

# LinearRegression.fit(x,y)


model.fit(x,y)# we r plotting the values of x,y in linear regression library
```

```python
#8.PREDICT THE MODEL
y_pred = model.predict(x)#using the i/p values , we predict the o/p
y_pred#PREDICTED OUTPUT VALUES
```

y #ACTUAL OUTPUT VALUES

```python
#conclusion:
#wE HAVE TO COMPARE THE corresponding values of y_pred and y.
#When we compare,we come to know that there is huge difference.
#This huge difference does not mean that our model has predicted wrong
#it only means that our model is NOT LINEAR/LESS LINEAR
#LINEARITY of a model depends SIZE and Nature of the data.
```

```python
#INDIVIDUAL PREDICTION
#I want to predict the price for 2000 sqft
model.predict([[2000]])
```
