```python
# Logic of KNN classification
# Choose the value of k (number of nearest neighbors)
# Plot all the data points in the coordinate system
# Calculate the distance from the data points to the predictive point
# Find the nearest points(k points) based on distance formula
# Check the labels/targets of the selected nearest points
# Based on majority of the labels, assign the label to the predictive point
```

```python
#1.Let us take the data and create dataframe
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/ameenmanna8824/DATASETS/refs/heads/main/height_weight_knn.csv')
df
```

```python
df.info()
```

```python
#inputs:featres
df.isnull().sum()
```

```python
# Inputs : Features
# Outputs: Labels/Targets
```

```python
X = df[['Weight(x2)','Height(y2)']].values
y = df['Class'].values
```

```python
# splitting of data into training and testing datasets
# from sklearn.model_selection import train_test_split
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model = KNeighborsClassifier(n_neighbors=5,metric='euclidean')
```

```python
model.fit(X,y)
```

```python
model.predict([[54.2,167.5]])
```

```python
import joblib
joblib.dump(model,'knn_model')
```

```python
%%writefile app.py
import streamlit as st
import joblib
model = joblib.load('knn_model')
st.header("BMI predictor Applet")
w = st.number_input("Enter the weight(in kg)",min_value=40,max_value=90,step=1)
h = st.number_input("Enter the height(in cm)",min_value=135,max_value=200,step=1)
if st.button("PREDICT CLASS"):
    op = model.predict([[w,h]])
    st.subheader(f"The person with above weight and height is classified as {op[0]}.")
```
