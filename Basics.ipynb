```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[5,6,7,8]
plt.title("hello")
plt.xlabel('X_axis')
plt.ylabel('Y_axis')
plt.plot(x,y,color='red',marker='*')
```

```python
overs=[0,4,8,12,16,20]
team1=[0,31,67,98,123,145]
team2=[0,27,75,103,110,135]
plt.title("anlaysis of cricket match")
plt.xlabel('overs')
plt.ylabel('runs')
plt.plot(overs,team1,color='green',marker='^')
plt.plot(overs,team2,color="red",marker='^')
plt.grid()
```

```python

plt.title("overs vs team2")
plt.xlabel('overs')
plt.ylabel('team2')
plt.plot(overs,team2,color='red',marker='*')
```

```python

plt.title("team1 vs team2")
plt.xlabel('team1')
plt.ylabel('team2')
plt.plot(team1,team2,color='red',marker='*')
```

```python
name=['a','b','c','d']
percent=[9,56,789,12]
plt.bar(name,percent,color=['red','green','violet'],ls='-',edgecolor='k')
plt.title('marks analysis')
plt.xlabel("names",color='red')
plt.ylabel('percentages',color='green')
```
